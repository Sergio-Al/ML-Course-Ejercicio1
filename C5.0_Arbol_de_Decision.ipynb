# C5.0 — Árbol de Decisión


---


## Descripción

**C5.0** (también llamado **See5** en su versión para Windows) es el **sucesor comercial de C4.5**,
desarrollado por el mismo autor, **J. Ross Quinlan** (RuleQuest Research). Es un algoritmo de
aprendizaje supervisado que induce **árboles de decisión** y **conjuntos de reglas `if-then`**
para tareas de **clasificación**.

Conserva el **mismo núcleo** que C4.5 —particionamiento recursivo *top-down* guiado por medidas
basadas en la **entropía de la información**— pero lo mejora de forma sustancial. Sus diferencias
clave respecto a C4.5 son:

- **Boosting integrado (*trials*):** construye **varios árboles** y los combina por votación
  (basado en *AdaBoost*), aumentando la exactitud. Es su característica estrella.
- **Winnowing (selección automática de atributos):** descarta de antemano los atributos poco
  útiles o ruidosos, produciendo modelos más simples y robustos.
- **Mucho más rápido y ligero:** entrena órdenes de magnitud más rápido que C4.5 y usa menos
  memoria.
- **Árboles más pequeños** con exactitud igual o superior.
- **Costos de clasificación variables:** permite penalizar más unos errores que otros
  (aprendizaje sensible al costo).
- **Pesos por caso (*case weights*)** y mejor manejo de valores faltantes.

En resumen: **C5.0 = C4.5 + boosting + selección de atributos + eficiencia**. Sigue siendo un
modelo **interpretable** ("caja blanca").

## Tipo de Modelo

- **Tipo — MÉTODO DE APRENDIZAJE:** Aprendizaje **supervisado** · **Clasificación**
  (particionamiento recursivo *greedy* + **boosting** opcional → convierte el árbol único en un
  **ensamble**).
- **Tipo — por PARÁMETROS:** Modelo **no paramétrico** (la complejidad del árbol/ensamble crece
  con los datos; no asume distribución fija).
- **Tipo — DATOS DE APRENDIZAJE:** Datos **etiquetados** (tabulares). Atributos **categóricos y/o
  continuos**; tolera **valores faltantes**; admite **pesos por caso** y **costos de error
  asimétricos**. Variable objetivo **discreta** (nominal).
- **Tipo — RESULTADO DEL ENTRENAMIENTO:** Un **árbol de decisión** o un **conjunto de reglas
  `if-then`**. Con *boosting* (`trials > 1`), el resultado es un **ensamble de árboles** que vota.


## Alg. Entrenamiento

- **C5.0** — inducción *top-down* de árboles de decisión (TDIDT) por particionamiento recursivo
  **voraz (greedy)** con criterio basado en la **ganancia/razón de ganancia** (herencia de C4.5),
  **poda** basada en error, y dos incorporaciones propias:
  ***boosting* adaptativo** (varios árboles combinados por votación) y **winnowing** (selección
  previa de atributos relevantes).


## Supuestos y Restricciones

**Supuestos**

- Los datos de entrenamiento son **representativos** de la población objetivo.
- Los atributos disponibles son **suficientes** para discriminar las clases.
- Las clases son **discretas y mutuamente excluyentes**.
- Los ejemplos se asumen aproximadamente **independientes**.

**Restricciones / limitaciones**

- El árbol base sigue siendo **voraz**: no garantiza el óptimo global (óptimos locales).
- Con *boosting* se **pierde parte de la interpretabilidad** (ya no es un solo árbol legible,
  sino muchos que votan).
- Genera **fronteras paralelas a los ejes** (*axis-parallel splits*).
- Fue **software propietario** de RuleQuest; solo la versión **monohilo** se liberó como código
  abierto (GPL). La implementación de referencia vive en **R** (paquete `C50`); en Python no hay
  una versión canónica ampliamente adoptada.
- Diseñado para **clasificación** (no para regresión).
- El *boosting* puede **sobreajustar** en datos con mucho ruido si se usan demasiados *trials*.

## Demostración práctica en Python

El núcleo de construcción del árbol es **idéntico al de C4.5** (ver la ficha de C4.5, sección
"desde cero" con *gain ratio*). Lo **nuevo** en C5.0 es el **boosting** y la **selección de
atributos**. Como en Python no existe una implementación canónica de C5.0, ilustramos esas dos
ideas con `scikit-learn` (un árbol con criterio de entropía + un *ensamble por boosting*), y al
final incluimos el **C5.0 real** ejecutable en R dentro de Colab.


### 1) La idea estrella de C5.0: *boosting* (varios árboles > un árbol)

````python
# scikit-learn y numpy ya vienen instalados en Colab.
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import AdaBoostClassifier
from sklearn.metrics import accuracy_score
import numpy as np

# Dataset real de clasificacion (tumor benigno / maligno)
X, y = load_breast_cancer(return_X_y=True)
X_tr, X_te, y_tr, y_te = train_test_split(
    X, y, test_size=0.30, random_state=42, stratify=y)

# (a) UN SOLO arbol -> nucleo tipo C4.5/C5.0 (criterio = entropia)
arbol = DecisionTreeClassifier(criterion="entropy", random_state=42)
arbol.fit(X_tr, y_tr)
acc_arbol = accuracy_score(y_te, arbol.predict(X_te))

# (b) BOOSTING -> muchos arboles que votan (idea de C5.0: parametro "trials")
boost = AdaBoostClassifier(
    estimator=DecisionTreeClassifier(criterion="entropy", max_depth=2, random_state=42),
    n_estimators=50, random_state=42)
boost.fit(X_tr, y_tr)
acc_boost = accuracy_score(y_te, boost.predict(X_te))

print(f"(a) Un solo arbol de decision : exactitud = {acc_arbol:.3f}")
print(f"(b) Boosting con 50 arboles   : exactitud = {acc_boost:.3f}")
print("\nEl boosting suele igualar o superar al arbol individual:")
print("esa es, en esencia, la mejora que aporta C5.0 con sus 'trials'.")
````

### 2) La otra idea: *winnowing* (el modelo ignora atributos inútiles)

C5.0 descarta automáticamente atributos poco informativos. Un árbol de decisión ya hace algo
parecido: solo usa los atributos que aportan información. Veámoslo con las **importancias**.


````python
from sklearn.datasets import load_breast_cancer
importancias = arbol.feature_importances_
usados = int(np.sum(importancias > 0))
print(f"Atributos disponibles      : {X.shape[1]}")
print(f"Atributos realmente usados : {usados}  (el resto fueron 'descartados')")

# Top 5 atributos mas relevantes
nombres = load_breast_cancer().feature_names
orden = np.argsort(importancias)[::-1][:5]
print("\nTop 5 atributos por importancia:")
for i in orden:
    print(f"  {nombres[i]:28s} {importancias[i]:.3f}")
````

### 3) El C5.0 *real*: ejecutar el paquete `C50` de R en Colab

La implementación de referencia de C5.0 (con código C del propio Quinlan) está en **R**. Colab
puede correr R desde una celda de Python gracias a `rpy2`. Ejecuta estas dos celdas en orden.


````python
# Activa la "magia" de R dentro de Colab (rpy2 viene preinstalado)
%load_ext rpy2.ipython
````

````python
%%R
# Instala el paquete C50 (solo la primera vez; puede tardar un poco)
if (!require("C50", quietly = TRUE)) {
  install.packages("C50", repos = "https://cloud.r-project.org")
}
library(C50)

data(iris)
set.seed(42)

# trials = 10  -> BOOSTING con 10 arboles (rasgo distintivo de C5.0)
# winnow = TRUE -> seleccion automatica de atributos
modelo <- C5.0(x = iris[, 1:4], y = iris$Species,
               trials = 10,
               control = C5.0Control(winnow = TRUE))

summary(modelo)   # muestra el arbol, atributos usados y el error
````
